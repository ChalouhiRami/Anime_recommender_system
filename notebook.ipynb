```python
import pandas as pd

df = pd.read_csv("anime-dataset-2025.csv")
df.head(3)
```

```python
print(df.shape)
print(df['Synopsis'].isna().sum())
print(df['Genres'].isna().sum())
print(df['Themes'].isna().sum())
```

```python
missing_synopsis = df['Synopsis'].isna()
print(df.loc[missing_synopsis, 'Genres'].isna().sum())

missing_synopsis_and_genres = missing_synopsis & df['Genres'].isna()
print(df.loc[missing_synopsis_and_genres, 'Themes'].isna().sum())
```

```python
embedding_text =[]

synopsis_genres=df['Synopsis'].combine_first(df['Genres'])
print(synopsis_genres.isna().sum())
synopsis_genres_themes=synopsis_genres.combine_first(df['Themes'])
print(synopsis_genres_themes.isna().sum())
```

```python
df['embedding_text'] = synopsis_genres_themes
df = df.dropna(subset=['embedding_text'])
print(df.shape)



```
